### Exercise 7. Challenge, no automatic check

Write a complete guessing game:

1. The computer picks a number between 1 and 100.
2. The player gets at most 7 guesses.
3. After each guess, print `Too small` or `Too large`.
4. If the player runs out of guesses, reveal the number.
5. Reject any input outside 1 to 100 without using up a guess.

Include a docstring and comments.

````python
# TODO: your program here
import random 
""" 
This game randomly selects a number between 1 and 100. 
The user has 7 attempts to guess the number. Inputs outside 1-100 
are rejected without penalty. 
""" 
secret_number = random.randint(1, 100) 
guesses_left = 7 
print("Guess the number between 1 and 100. You have 7 attempts.") 
while guesses_left > 0: 
    guess = int(input("Enter your guess: ")) 
if guess < 1 or guess > 100: 
    print("Out of bounds! Guess between 1 and 100.") 
    continue  # Skips penalty and restarts loop 
guesses_left -= 1 
if guess == secret_number: 
    print("Correct! You won.") 
    break 
elif guess < secret_number: 
    print("Too small.") 
else: 
    print("Too large.")
if guesses_left == 0 and guess != secret_number: 
    print(f"Out of guesses. The number was {secret_number}.")
````
